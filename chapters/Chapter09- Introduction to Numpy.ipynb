## Example
Let's assume we have a matrix where each row corresponds to a gene and the columns correspond to different samples.
1. Calculate the expression changes for each gene:
2. Find the genes that have the most changes:
3. Display these genes:
4. Display the names of the genes if you have an array of names:


```python
# Extract gene names from the first row
with open("./samples/CellBench-lung3k_10x_5cl_data.csv") as f:
    header = f.readline().strip().split(",")
    gene_names = np.array(header[1:])  # Remove the first column (cell IDs)
    N = len(header)

import numpy as np

# Load numerical gene expression data
expression = np.loadtxt("./samples/CellBench-lung3k_10x_5cl_data.csv", 
                        delimiter=",", 
                        skiprows=1, 
                        usecols=range(1, N))  # Use columns 2 to N (gene expression values)

print("Shape of data matrix:", expression.shape)

# 1. Calculate expression difference for each gene (max - min across all cells)
gene_diff = np.max(expression, axis=0) - np.min(expression, axis=0)
print(gene_diff)

# 2. Get indices of top 5 genes with highest expression variation
#Returns the indices that would sort an array.
top_indices = np.argsort(gene_diff)[-5:]
print("Top gene indices:", top_indices)

# 3. Get expression values of those top genes across all cells
print("Expression of top genes:\n", expression[:, top_indices])

# 4. Print names of the top 5 most variable genes
print("Top gene names:", gene_names[top_indices])
```

### example 1
Imagine you’re running a lab with 100 mice. Each one’s been on a different diet. You want to see who got fatter and who shrank into fur and bones.With NumPy, in just 5 lines of code you can load all their weights, compute weight changes, identify who got chunkier, and maybe even send some mice back to normal chow before they disappear!You can even make a plot and label it ‘Mouse_42 on Keto.

```python
import numpy as np
import matplotlib.pyplot as plt

# Initial and final weights (100 mice)
np.random.seed(42)
initial_weights = np.random.normal(loc=25, scale=5, size=100)  # e.g., grams
final_weights = initial_weights + np.random.normal(loc=0, scale=3, size=100)

# Weight difference
delta = final_weights - initial_weights

# Identify mice who gained weight
gained_weight = delta > 1.0  # Threshold for noticeable gain

print("Mice who gained weight:", np.sum(gained_weight))

# Optional: visualize
plt.hist(delta, bins=20, color='lightblue', edgecolor='black')
plt.axvline(x=0, color='red', linestyle='--', label='No change')
plt.xlabel("Weight Change (g)")
plt.ylabel("Number of mice")
plt.title("Mouse Diet Effect (Mouse_42 on Keto?)")
plt.legend()
plt.show()
```

### Example 2

you have gene expression data for 2,000 genes across 100 cells. You want to find which genes change the most between two groups — for example, healthy vs. cancerous cells.With NumPy, you can:🔹 Load the matrix from CSV🔹 Compute the mean expression per gene per group🔹 Calculate differences🔹 Use np.argsort to find the top 10 most changed genes.This is exactly what happens in RNA-seq differential expression analysis!


```python
import numpy as np

# Simulated gene expression: 2000 genes × 100 cells
np.random.seed(1)
gene_expression = np.random.rand(2000, 100) * 100

# First 50 cells: healthy, last 50: cancer
healthy = gene_expression[:, :50]
cancer = gene_expression[:, 50:]

# Compute average expression per gene in each group
mean_healthy = np.mean(healthy, axis=1)
mean_cancer = np.mean(cancer, axis=1)

# Difference in expression
diff = mean_cancer - mean_healthy

# Get top 10 most changed genes (by absolute value)
top10_genes = np.argsort(np.abs(diff))[-10:][::-1]

print("Top 10 genes with largest expression change:")
print(top10_genes)
print("Differences:", diff[top10_genes])
```
