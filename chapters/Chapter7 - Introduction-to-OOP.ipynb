- Inheritance : Share behavior/data from one class to another

## Encapsulation
- Encapsulation :  Hiding internal details, exposing only what’s needed

```python
class Cell:
    def __init__(self):
        self._energy = 100  # Protected (by convention)

    def use_energy(self, amount):
        if amount <= self._energy:
            self._energy -= amount
        else:
            print("Not enough energy")

    def get_energy(self):
        return self._energy
```

Underscore _ is a Python convention for "internal use".
Encapsulation helps protect internal state.

```python
c = Cell()
print(c.get_energy())
```

```python
c.use_energy(20)
```

```python
c._energy = 1000
```

### Class Variables vs Instance Variables
- Class variables are shared across all instances:


```python
class Sequence:
    transcription_table = {'A': 'U', 'T': 'A', 'C': 'G', 'G': 'C'}
    enz_dict = {'EcoRI': 'GAATTC', 'EcoRV': 'GATATC'}

    def __init__(self, seqstring):
        self.seqstring = seqstring.upper()

    def transcription(self):
        result = ""
        for letter in self.seqstring:
            if letter in 'ATCG':
                result += self.transcription_table[letter]
        return result

    def restriction(self, enz):
        try:
            enz_target = self.enz_dict[enz]
            return self.seqstring.count(enz_target)
        except KeyError:
            return 0
```

## 🧬 Sequence Class – A Biology Example

```python
class Sequence:
    # 🔒 Class-level translation table (shared by all instances)
    transcription_table = {'A': 'U', 'T': 'A', 'C': 'G', 'G': 'C'}
    
    def __init__(self, seqstring):
        # 📥.self: Instance-specific DNA sequence (converted to uppercase)
        self.seqstring = seqstring.upper()

    def transcription(self):
        result = ""  # 🧪 Local variable used only inside this method
        for letter in self.seqstring:
            result += self.transcription_table[letter]  # 🔄 Lookup using the class-level table via self
        return result
        
#Sequence.transcription_table  ✅  # Correct: accessing the class variable directly through the class
#self.transcription_table      ✅  # Also correct and more common inside class methods
#transcription_table           ❌  # Error: not defined in the local scope of the method
```

```python
# ✅ Why use self.transcription_table instead of just transcription_table?
# Although transcription_table is defined at the class level, using self.transcription_table inside methods:
# - Allows Python to first check if the instance has a customized version (e.g., s.transcription_table).
# - Falls back to the class version if the instance doesn't override it (e.g., Sequence.transcription_table).
# - Keeps the code flexible and consistent with object-oriented practices (e.g., easier to modify instance-specific behavior).

# Example:
# s1 = Sequence("ATCG")
# s1.transcription_table = {'A': 'X', 'T': 'Y', 'C': 'Z', 'G': 'W'}  # Custom version for s1
# print(s1.transcription())  # Uses X Y Z W for s1

# Example of class-level modification:
# Sequence.transcription_table = {'A': '-', 'T': '-', 'C': '-', 'G': '-'}
# s2 = Sequence("ATCG")
# print(s2.transcription())  # Uses - - - - for all future instances like s2
```

## Plasmid Class (inherits from Sequence)
- A Plasmid is a special type of DNA sequence. A plasmid is a small circular DNA molecule found in bacteria and some other microscopic organisms.

Let's build a Plasmid class that inherits from a Sequence class and adds antibiotic resistance features.



```python
class Plasmid(Sequence):
    ab_res_dict = {'Tet': 'ctagcat', 'Amp': 'CACTACTG'}
    def __init__(self, seqstring):
        super().__init__(seqstring)
    def ab_res(self, ab):
        return self.ab_res_dict.get(ab, '') in self.seqstring
```

```python
bacteria_sample = Plasmid('ctagcatatgatatcggagaggctagcatatatcggtgtcaaCACTACTGctagcat')
print(bacteria_sample.ab_res('Amp'))
```

Key Methods:.
complement() ➔ Returns complement strand.

```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction as gc

# Create a sequence object
first_seq = Seq("GCTATGCAGC")

# Complement and reverse complement
print(first_seq.complement())          # CGATACGTCG
print(first_seq.reverse_complement())  # GCTGCATAGC

# Convert to string
print(str(first_seq))  # GCTATGCAGC

# Indexing and slicing
print(first_seq[:5])   # GCTAT
print(len(first_seq))  # 10
print(first_seq[0])    # G
```

###  __ iter__ and __ next__
- Allows objects to be iterated with a for loop.
for calls iter(object) ➔ triggers __iter__() ➔ uses __next__() to fetch items.


```python
class Straight:
    def __init__(self, data):
        self.data = data
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index == len(self.data):
            raise StopIteration
        answer = self.data[self.index]
        self.index += 1
        return answer
```

```python
a = Straight('1235689')
for i in a:
    print(i)
```

## Example: Sequence Class with Special Methods

```python
# Sequence class with special methods

class Sequence:
    # Transcription table for DNA to RNA
    transcription_table = {'A': 'U', 'T': 'A', 'C': 'G', 'G': 'C'}
    # Complement table for DNA complement strand
    comp_table = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}

    def __init__(self, seqstring):
        # Store the sequence in uppercase
        self.seqstring = seqstring.upper()

    def restriction(self, enz):
        # Dictionary of enzyme recognition sites
        enz_dict = {'EcoRI': 'ACTGG', 'EcoRV': 'AGTGC'}
        try:
            target = enz_dict[enz]
        except KeyError:
            raise ValueError('No such enzyme in our enzyme DB')
        # Count occurrences of target sequence
        return self.seqstring.count(target)

    def __getitem__(self, index):
        # Return the character at the given index
        return self.seqstring[index]

    def __len__(self):
        # Return the length of the sequence
        return len(self.seqstring)

    def __str__(self):
        # String representation of sequence, abbreviated if long
        if len(self.seqstring) >= 28:
            return '{0}...{1}'.format(self.seqstring[:25], self.seqstring[-3:])
        else:
            return self.seqstring

    def transcription(self):
        # Transcribe DNA to RNA
        tt = ''
        for x in self.seqstring:
            if x in 'ATCG':
                tt += self.transcription_table[x]
        return tt

    def complement(self):
        # Get the complement strand
        tt = ''
        for x in self.seqstring:
            if x in 'ATCG':
                tt += self.comp_table[x]
        return tt
```

```python
seq1 = Sequence('ATGCGTACGTTAGC')
print(seq1.transcription())
```

```python
print(seq1.complement())
```

## Fun & Practical Example
- think of OOP in Python like the world of cells in biology.
Each cell is an object with attributes like DNA, size, type, and methods like dividing or apoptosis!
You can create a class called Cell, and then make different cells like muscle cells, neurons, or immune cells by inheritance.
If an ImmuneCell encounters a virus, it runs its attack() method! But if it's food, it just runs ignore()!
Just like how we modified the dictionary in Zdic, in OOP we can give new behaviors to objects.


```python
class Cell:
    def __init__(self, cell_type):
        self.cell_type = cell_type  # e.g., "immune", "neuron"

    def react(self, stimulus):
        if self.cell_type == "immune" and stimulus == "virus":
            return "Attack the virus! 💥"
        elif stimulus == "food":
            return "Ignore, it's food 🍏"
        else:
            return "Stay calm 😎"

t_cell = Cell("immune")
print(t_cell.react("virus"))  # Attack the virus! 💥
print(t_cell.react("food"))   # Ignore, it's food 🍏
```
