Instructions:
1. Load the CSV file.
2. Compute mean expression per group.
3. Plot the result as a bar chart.
4. Apply Z-score normalization.
5. Identify genes with higher expression in the treatment group.

0. gene_expression.csv:

```python
import pandas as pd

data = {
    "SampleID": ["S1", "S2", "S3", "S4", "S5"],
    "Group": ["Control", "Treatment", "Control", "Treatment", "Control"],
    "GeneA": [2.3, 5.6, 1.2, 4.8, 3.1],
    "GeneB": [5.1, 4.2, 3.9, 6.5, 4.3],
    "GeneC": [3.3, 6.1, 2.7, 5.4, 3.8]
}

df = pd.DataFrame(data)

df.to_csv("my_gene_expression.csv", index=False)

df
```

Step 1 — Load the CSV file

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the gene expression dataset
df = pd.read_csv("my_gene_expression.csv")
# Add groups
# Display the first five rows
print(df.head())
```

Step 3 — Compute mean expression per group
- What is the average expression of each gene in Control and Treatment groups?

```python
# Calculate the mean expression of each gene for each group
group_means = df.groupby("Group")[["GeneA", "GeneB", "GeneC"]].mean()

print(group_means)
```

Step 4 — Plot the results

```python
# Create a bar chart of mean gene expression by group
group_means.T.plot(kind="bar", figsize=(8, 5))

# Add plot labels and title
plt.xlabel("Gene")
plt.ylabel("Mean Expression")
plt.title("Mean Gene Expression: Control vs Treatment")

# Display the legend
plt.legend(title="Group")

# Improve the layout
plt.tight_layout()

# Display the plot
plt.show()
```

Step 5 — Apply Z-score normalization

Z-score tells us how far a value is from the average, measured in standard deviations.

- x = expression value
- μ = mean expression
- σ = standard deviation

```python
# Select only gene expression columns
gene_columns = ["GeneA", "GeneB", "GeneC"]

# Calculate Z-score for each gene across all samples
z_scores = (
    df[gene_columns] - df[gene_columns].mean()
) / df[gene_columns].std()

print(z_scores)
```

```python
# Create a copy of the original dataset
df_zscore = df.copy()

# Add Z-score values for each gene
df_zscore[gene_columns] = z_scores

# Display the normalized dataset
print(df_zscore)
```

Step 6 — Identify genes with higher expression in Treatment
- Which genes have higher average expression in the Treatment group?

```python
group_means = df.groupby("Group")[gene_columns].mean()

print(group_means)
```

```python
# Calculate the difference between Treatment and Control means
difference = (
    group_means.loc["Treatment"] -
    group_means.loc["Control"]
)

print("Treatment - Control:")
print(difference)
```

- GeneA → higher in Treatment
- GeneB → lower in Treatment
- GeneC → higher in Treatment

Step 7 — Automatically identify the higher-expression genes

```python
# Select genes with higher mean expression in the Treatment group
higher_in_treatment = difference[difference > 0]

print("Genes with higher expression in Treatment:")
print(higher_in_treatment)
```

GeneA  exhibit higher mean expression levels in the Treatment group compared to the Control group.

- Full version of the code:

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the gene expression dataset
df = pd.read_csv("my_gene_expression.csv")

# Display the first five rows
print("First five rows:")
print(df.head())

# Display the dataset shape
print("\nDataset shape:")
print(df.shape)

# Define the gene columns
gene_columns = ["GeneA", "GeneB", "GeneC"]

# Calculate mean expression for each group
group_means = df.groupby("Group")[gene_columns].mean()

print("\nMean expression by group:")
print(group_means)

# Plot mean gene expression by group
group_means.T.plot(kind="bar", figsize=(8, 5))

# Add plot labels and title
plt.xlabel("Gene")
plt.ylabel("Mean Expression")
plt.title("Mean Gene Expression: Control vs Treatment")
plt.legend(title="Group")
plt.tight_layout()
plt.show()

# Calculate Z-score normalization for each gene
z_scores = (
    df[gene_columns] - df[gene_columns].mean()
) / df[gene_columns].std()

print("\nZ-score normalized data:")
print(z_scores)

# Calculate the difference between Treatment and Control means
difference = (
    group_means.loc["Treatment"] -
    group_means.loc["Control"]
)

print("\nTreatment - Control difference:")
print(difference)

# Identify genes with higher mean expression in Treatment
higher_in_treatment = difference[difference > 0]

print("\nGenes with higher expression in Treatment:")
print(higher_in_treatment)
```

### Example: Gene Expression in Sleep-Deprived Students

Objective: Investigate the difference in the mean expression level of a melatonin receptor gene, such as MTNR1B, between two hypothetical groups:

- Group 1: Students who are sleep deprived
- Group 2: Students who get sufficient sleep

The goal is to use Pandas to compare the average gene expression between the two groups.


Let's use a small biological example. Suppose we want to investigate whether sleep status is associated with the expression of a gene involved in circadian biology. These data are completely fictional and are only for learning Pandas.

```python
import pandas as pd

# Create a fictional biological dataset
data = {
    "student": ["Ali", "Zahra", "Omid", "Fatemeh", "Reza", "Sara"],
    "sleep_status": [
        "deprived", "normal", "deprived",
        "normal", "deprived", "normal"
    ],
    "MTNR1B_expression": [2.1, 5.5, 1.8, 6.0, 2.5, 5.8]
}

df = pd.DataFrame(data)

df
```

```python
# Calculate mean expression for each sleep group
mean_expression = (
    df.groupby("sleep_status")["MTNR1B_expression"]
      .mean()
)

print(mean_expression)
```

```python
df
```

What did we just do? We grouped our samples according to sleep status, selected the gene expression column, and calculated the mean for each group.

```python
# Plot mean gene expression for each group
mean_expression.plot(
    kind="bar",
    title="Mean MTNR1B Expression by Sleep Status"
)
```

In Pandas, groupby() means: split the data into groups, perform an operation on each group, and combine the results.

The graph shows a difference in our fictional dataset, but a difference in group means does not automatically mean a biological effect. Real biological conclusions require appropriate experimental design, sufficient samples, statistical testing, and biological validation.
