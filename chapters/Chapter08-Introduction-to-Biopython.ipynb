### 4- Align
The Align module in Biopython handles multiple sequence alignments.
The central object is MultipleSeqAlignment, which stores already aligned sequences.
- It does not perform the alignment
— it only stores aligned sequences


```python
from Bio.Align import MultipleSeqAlignment
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord

# Define sequences
seq1 = 'MHQAIFIYQIGYPLKSGYIQSIRSPEYDNW'
seq2 = 'MH--IFIYQIGYALKSGYIQSIRSPEY-NW'

# Create sequence records
seq_rec_1 = SeqRecord(Seq(seq1), id='asp')
seq_rec_2 = SeqRecord(Seq(seq2), id='unk')

# Create the alignment object
align = MultipleSeqAlignment([seq_rec_1, seq_rec_2])

print(align)
```

- Adding a New Sequence

- Parsing BLAST Results in Biopython

```python
from Bio.Blast import NCBIXML

with open("./samples/testblast.xml") as result_handle:
    blast_records = NCBIXML.parse(result_handle)
    for blast_record in blast_records:
        print("Query:", blast_record.query)
        for alignment in blast_record.alignments:
            print("  Match:", alignment.hit_def)
            for hsp in alignment.hsps:
                print("    Score:", hsp.score)
                print("    E-value:", hsp.expect)
                print("    Identity:", hsp.identities)
                print("    Alignment length:", hsp.align_length)
                print("    Query sequence:", hsp.query)
                print("    Match sequence:", hsp.sbjct)
```

- example: Reading BLAST output with Biopython

```python
from Bio.Blast import NCBIXML

# Open the BLAST XML output file
with open("./samples/blast_output.xml") as result_handle:
    blast_record = NCBIXML.read(result_handle)  # Use .read() for single-query files

# Iterate through each alignment and its HSPs
for alignment in blast_record.alignments:
    print("Hit ID:", alignment.hit_id)
    for hsp in alignment.hsps:
        # Print basic info from each HSP
        print("Score:", hsp.score)
        print("E-value:", hsp.expect)
        print("Query start:", hsp.query_start)
        print("Subject start:", hsp.sbjct_start)
        print("Query sequence:", hsp.query)
        print("Subject sequence:", hsp.sbjct)
```

- Protein weight calculator with Biopython

No need to create your own amino acid dictionary — Biopython includes it!


```python
from Bio.Data.IUPACData import protein_weights as pw

# Get protein sequence from user
protseq = input('Enter your protein sequence: ')

total_w = 0
for aa in protseq:
    total_w += pw.get(aa.upper(), 0)

# Subtract weight of water for each peptide bond
total_w -= 18 * (len(protseq) - 1)

print(f'The net weight is: {total_w}')
```

What is eUtils
- Server-side API to query Entrez databases programmatically
- Returns data mostly in XML
- URLs are built by scripts, not manually

- Code Example


```python
# Import the Entrez module from Biopython
from Bio import Entrez

# Set your email address (required by NCBI)
Entrez.email = "parissa.ahmadi92@gmail.com"

# Define the database to search
db = "pubmed"

# Perform a search using esearch from eUtils
with Entrez.esearch(db=db, term="rna sequencing", retmax=5) as h_search:
    # Parse the search results
    record = Entrez.read(h_search)

# Retrieve the list of IDs from the search results
res_ids = record["IdList"]

# For each ID, fetch the summary information
for r_id in res_ids:
    with Entrez.esummary(db=db, id=r_id) as h_summ:
        # Parse the summary information
        summ = Entrez.read(h_summ)
        # Print the title and DOI of the article
        print(summ[0]["Title"])
        print(summ[0].get("DOI", "DOI not available"))
        print("=" * 50)
```

```python
#eUtils
from Bio import Entrez
my_em = "parissa.ahmadi92@gmail.com"
db = "gene"
term = "cobalamin synthase homo sapiens"
h_search = Entrez.esearch(db=db, email=my_em, term=term)
record = Entrez.read(h_search)
res_ids = record["IdList"]
for r_id in res_ids:
    h_summ = Entrez.esummary(db=db, id=r_id, email=my_em)
    summ = Entrez.read(h_summ)
    print(r_id)
    gene_info = summ['DocumentSummarySet']['DocumentSummary'][0]
    print('Gene Name:',gene_info['Name'])
    print('Description:',gene_info['Description'])
    print('Summary:',gene_info['Summary'])
    print('==============================================')
```

### 14- Restriction Enzymes
Recombinant DNA technology allows the combination of DNA sequences from different organisms that would not normally occur together. This process is accomplished by restriction endonucleases, enzymes that act as specific molecular scissors

- Example: #EcoRI ---> GAATTC

- CGCGAATTCGCG
- GCGCTTAAGCGC

- The restriction site is located in the middle of the top strand:
- CGC-GAATTC-GCG

The resulting pieces after cutting:
- CGC GAATTCGCG
- GCGCTTAA GCGC



- Accessing Enzyme Information:

```python
from Bio import Restriction
Restriction.EcoRI
```

### 16- Sequencing
DNA sequencing trace data is read by the Phred program. This program calls bases, assigns quality values to them, and writes both the base calls and quality values into output files with the .phd.1 extension.

```python
# Reading the file lines
with open('./samples/phd1') as fh:
    lines = fh.readlines()

# Parse the sequence and comments from the lines
sequence = []
comments = []
in_dna = False
in_comment = False

for line in lines:
    line = line.strip()
    if line.startswith('BEGIN_SEQUENCE'):
        continue
    elif line.startswith('BEGIN_COMMENT'):
        in_comment = True
        continue
    elif line.startswith('END_COMMENT'):
        in_comment = False
        continue
    elif line.startswith('BEGIN_DNA'):
        in_dna = True
        continue
    elif line.startswith('END_DNA'):
        in_dna = False
        continue

    if in_comment:
        comments.append(line)
    elif in_dna:
        # Extracting DNA sequence (only the first character of each line)
        sequence.append(line.split()[0])

# Join sequence into one string
sequence_str = ''.join(sequence)

# Output the results
print("Comments:")
for comment in comments:
    print(comment)

print("\nDNA Sequence:")
print(sequence_str)
```

### Example 3-2
Write a version that gets sequences from NCBI, uses a specific reference, performs alignment, and reports mutations A→G, C→T, etc. in a table

```python
from Bio import Entrez, SeqIO
from Bio.Align import PairwiseAligner
import pandas as pd

# ---------------------------------------------------------
# 1. NCBI settings
# ---------------------------------------------------------

# Replace this with your real email address
Entrez.email = "your_email@example.com"

# ---------------------------------------------------------
# 2. Define the reference sequence
# ---------------------------------------------------------

# Example: NCBI accession.version of the reference sequence
REFERENCE_ID = "REFERENCE_ACCESSION"

# Fetch the reference sequence from NCBI
handle = Entrez.efetch(
    db="nucleotide",
    id=REFERENCE_ID,
    rettype="fasta",
    retmode="text"
)

reference = SeqIO.read(handle, "fasta")
handle.close()

print("Reference:")
print(reference.id)
print(reference.description)
print("Length:", len(reference.seq))


# ---------------------------------------------------------
# 3. Search NCBI for sample sequences
# ---------------------------------------------------------

search_term = (
    '"Influenza A virus"[Organism] '
    'AND hemagglutinin[Gene] '
    'AND biomol_mrna[PROP]'
)

handle = Entrez.esearch(
    db="nucleotide",
    term=search_term,
    retmax=10
)

search_result = Entrez.read(handle)
handle.close()

sample_ids = search_result["IdList"]

print("\nNumber of samples found:", len(sample_ids))


# ---------------------------------------------------------
# 4. Download the sample sequences
# ---------------------------------------------------------

if not sample_ids:
    raise RuntimeError("No sequences were found.")

handle = Entrez.efetch(
    db="nucleotide",
    id=sample_ids,
    rettype="fasta",
    retmode="text"
)

samples = list(SeqIO.parse(handle, "fasta"))
handle.close()

print("Downloaded samples:", len(samples))


# ---------------------------------------------------------
# 5. Create the pairwise aligner
# ---------------------------------------------------------

aligner = PairwiseAligner()

# Use a global alignment
aligner.mode = "global"


# ---------------------------------------------------------
# 6. Compare each sample with the reference
# ---------------------------------------------------------

mutation_results = []

for sample in samples:

    print("\nComparing:")
    print("Reference:", reference.id)
    print("Sample:   ", sample.id)

    # Perform pairwise alignment
    alignment = aligner.align(
        reference.seq,
        sample.seq
    )[0]

    # Extract aligned sequences
    ref_aligned = alignment[0]
    sample_aligned = alignment[1]

    # Reference coordinate
    ref_position = 0

    # Compare aligned sequences
    for ref_base, sample_base in zip(ref_aligned, sample_aligned):

        # Move along the reference sequence
        if ref_base != "-":
            ref_position += 1

        # Ignore gaps for this simple SNP analysis
        if ref_base == "-" or sample_base == "-":
            continue

        # Detect nucleotide differences
        if ref_base != sample_base:

            mutation_results.append({
                "Sample": sample.id,
                "Reference": reference.id,
                "Position": ref_position,
                "Reference_Base": ref_base,
                "Sample_Base": sample_base,
                "Mutation": f"{ref_base}→{sample_base}"
            })


# ---------------------------------------------------------
# 7. Create a table of detected mutations
# ---------------------------------------------------------

mutation_table = pd.DataFrame(mutation_results)

if mutation_table.empty:

    print("\nNo nucleotide substitutions were detected.")

else:

    print("\nDetected mutations:")
    print(mutation_table.to_string(index=False))

    # Save the results
    mutation_table.to_csv(
        "influenza_mutations.csv",
        index=False
    )

    print("\nResults saved to: influenza_mutations.csv")

```
